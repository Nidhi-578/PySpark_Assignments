3. What are the main components of Kafka?

The main components of Kafka are Producers, Topics, Partitions, Brokers, Consumers, Consumer Groups, Offsets, and Replicas. Producers publish events to topics, and topics are divided into partitions for scalability and parallel processing. Brokers store these partitions and their replicas. Consumers read events from partitions, and consumer groups allow multiple consumers to process partitions in parallel. Offsets track the consumer's position within a partition, while replication provides fault tolerance and high availability. Together, these components allow Kafka to provide scalable, reliable, and high-throughput event streaming

4. What is a Kafka topic?

A Kafka topic is a named logical stream to which producers publish events and from which consumers consume events. A topic is divided into partitions for horizontal scalability and parallel processing. Each partition maintains an ordered sequence of records identified by offsets. Kafka retains records based on its retention policy rather than deleting them immediately after consumption, which allows consumers to replay or reprocess data when required.

5. What is a partition in Kafka?

A partition is an ordered, immutable sequence of records within a Kafka topic. Kafka divides a topic into multiple partitions to achieve horizontal scalability and parallel processing. Partitions can be distributed across brokers, and consumers within a consumer group process different partitions in parallel. Each record has an offset within its partition, and Kafka guarantees ordering within a partition, but not across partitions. When a message has a key, Kafka typically uses the key to determine the partition, ensuring records with the same key are routed to the same partition.

6. Why does Kafka use partitions?

This allows multiple consumers within a consumer group to process partitions in parallel and enables Kafka to handle high-throughput workloads. Partitions also provide ordering at the partition level. When a message key is used, records with the same key are typically routed to the same partition, allowing ordering to be maintained for that key.

7. What is a broker?

Brokers can also host replicas of partitions, which provides fault tolerance and high availability. In a cluster, one broker may act as the leader for a partition while other brokers maintain replicas, and producers and consumers interact with the leader for that partition.

12. What happens when you increase the number of partitions of a topic?

When we increase the number of partitions of a Kafka topic, Kafka creates additional partitions, which increases the potential parallelism and allows more consumers within a consumer group to process the topic concurrently. Existing records are not automatically redistributed to the new partitions; they remain in their original partitions. New records can be distributed across the expanded partition set according to the producer's partitioning logic. We also need to be careful with key-based ordering because changing the partition count can change partition assignment for keys. 13. What is replication in Kafka?

Kafka replication is the mechanism by which Kafka maintains multiple copies of each partition across different brokers. One replica acts as the leader and the others act as followers. Producers and consumers interact with the partition leader, while followers replicate the leader's data. If the leader broker fails, Kafka can elect an eligible follower as the new leader. Replication therefore provides fault tolerance, high availability, and improved data durability. The number of copies is controlled by the replication factor

14. What is a replication factor?

Replication factor is the number of replicas maintained for each Kafka partition across different brokers. For example, with a replication factor of 3, each partition has three copies, typically one leader and two follower replicas. The leader handles normal produce and fetch operations, while followers replicate the leader's data. If the leader broker fails, an eligible in-sync replica can be elected as the new leader. Replication provides fault tolerance, high availability, and data durability.

15. Why is replication important?

Replication is important in Kafka because it provides fault tolerance, high availability, and data durability. Kafka maintains multiple replicas of each partition across different brokers. One replica acts as the leader and handles normal client operations, while follower replicas replicate the data. If the leader broker fails, an eligible in-sync replica can be elected as the new leader, allowing the partition to remain available and reducing the risk of data loss

16. What happens if a Kafka broker goes down?

When a Kafka broker goes down, Kafka detects the failure and identifies the partitions for which that broker was the leader. If replicas are available, Kafka elects an eligible in-sync replica as the new leader. Producers and consumers refresh their metadata and redirect requests to the new leader. When the failed broker recovers, its replicas catch up with the current leader before becoming fully in sync again. This failover mechanism, combined with replication and ISR, provides Kafka with high availability and fault tolerance.

17.What are follower replicas?

Follower replicas are copies of a Kafka partition maintained on other brokers. They continuously replicate records from the partition leader. The leader handles normal client operations, while followers maintain synchronized copies and provide redundancy. If the leader broker fails, an eligible in-sync follower can be elected as the new leader. Therefore, follower replicas are essential for Kafka's fault tolerance, high availability, and data durability.

18. What is the difference between Kafka and a traditional message queue?


The main difference is that Kafka is designed as a distributed event-streaming platform, whereas traditional message queues are primarily designed for asynchronous message or task delivery. Kafka persists events for a configurable retention period and tracks consumer position using offsets, so consumers can replay and independently process historical events. Kafka also supports multiple consumer groups consuming the same event stream and uses partitions for horizontal scalability and parallel processing. Traditional queues generally follow a consume-and-acknowledge model where messages are removed or become unavailable after successful processing. Therefore, I would typically use Kafka for high-throughput streaming and data pipelines, while a traditional message queue is more suitable for asynchronous task processing.

19. What is a Kafka producer?

A Kafka producer is an application or client that publishes records to Kafka topics. The producer serializes the record, determines the target partition using the configured partitioning strategy, and sends the record to the appropriate Kafka broker. Producers can use message keys to control partition assignment and maintain ordering for a key. They also support batching, compression, retries, and acknowledgment settings such as acks=0, acks=1, and acks=all to balance throughput and durability.

20. How does a producer send data to Kafka?

When a producer sends data to Kafka, the application creates a record containing a key and value. The producer serializes the record and uses its partitioning logic to determine the target partition. It obtains metadata about the topic and sends the record to the broker that is currently the leader for that partition. The leader appends the record to the partition log and replicates it to follower replicas according to the replication configuration. Finally, the broker sends an acknowledgment back to the producer based on the configured acks setting.

21. How does a producer decide which partition receives a message?

The Kafka producer uses a partitioner to determine which partition receives a record. If a key is provided, the partitioner uses the key to determine the partition, typically ensuring that records with the same key are routed to the same partition. This is important for maintaining ordering for that key. If no key is provided, the producer uses its partitioning strategy to distribute records across the available partitions. After determining the partition, the producer uses Kafka metadata to identify the broker that is currently the leader for that partition and sends the record there.

22. What is a partition key?

A partition key is a value in a Kafka record that the producer's partitioner uses to determine the target partition. For example, if we use customer_id as the key, records for the same customer are typically routed to the same partition. This is important when we need ordering for a particular entity, because Kafka guarantees ordering within a partition. The choice of partition key also affects data distribution and parallelism, so we should choose a key with good cardinality and relatively even distribution to avoid partition hotspots.

23. Why is a key important when producing Kafka messages?

A Kafka message key is important because it influences partition assignment. When we provide a key, the producer's partitioner typically routes records with the same key to the same partition. This allows us to maintain ordering for a particular business entity, such as a customer, account, or order. At the same time, the key affects how evenly data is distributed across partitions, so choosing a high-cardinality and well-distributed key is important to avoid hot partitions and maintain good parallelism.

24. What happens if the producer doesn't provide a key?

If the producer doesn't provide a key, the producer's partitioner uses its configured/default strategy to select a partition. Since there is no key to consistently associate related records with a particular partition, records can be distributed across multiple partitions. This can provide good load distribution and parallelism, but we cannot rely on topic-wide ordering. If ordering is required for a business entity, such as a customer or account, I would provide an appropriate key.

25. If you send records with the same key, will they always go to the same partition?

With the default key-based partitioning behavior, records with the same key are typically routed to the same partition for a given topic, as long as the partition count and partitioning strategy remain unchanged. This provides ordering for that key. However, it is not an absolute guarantee forever—if the number of partitions changes or a custom partitioner is used, the same key may map to a different partition. Therefore, if key-based ordering is important, partition-count changes should be carefully planned.

26. What is producer batching?

Producer batching in Kafka means accumulating multiple records for the same partition and sending them together in a request. It improves throughput and reduces network overhead. Two important configurations are batch.size, which controls the maximum batch size in bytes, and linger.ms, which controls how long the producer may wait for additional records before sending the batch. The producer sends the batch when it becomes full or when the linger time expires. However, increasing batching can introduce some latency, so it should be balanced according to the application's throughput and latency requirements.

27. What is a Kafka consumer?

A Kafka consumer is an application or client that subscribes to Kafka topics and reads and processes records from their partitions. Consumers track their processing position using offsets, and multiple consumers can work together in a consumer group, where Kafka assigns partitions among them for parallel processing. If a consumer fails, Kafka can rebalance the consumer group and reassign its partitions to another consumer. Consumers can also commit offsets so that processing can resume from a known position after a restart.

28. How does a consumer read messages from Kafka?

A Kafka consumer first subscribes to one or more topics. If it belongs to a consumer group, Kafka assigns partitions to the consumers in that group. The consumer obtains metadata about its assigned partitions and sends fetch requests to the brokers that are leaders for those partitions. It receives records in batches, processes them, and commits its offsets to record its progress. If a consumer fails, Kafka rebalances the consumer group and another consumer can take over its partitions, typically continuing from the last committed offset.

29. What is an offset?

An offset is a sequential identifier assigned to each record within a Kafka partition. It represents the record's position in that partition and is used by consumers to track their processing progress. Kafka stores consumer-group offsets so that consumers can resume processing after a restart or failure. Since offsets are partition-specific, the same offset number can exist in multiple partitions. Because Kafka retains records independently of consumer progress, consumers can also reset their offsets and replay or reprocess historical data when required.

30. Why are offsets important?

Offsets are important because they represent a consumer's position within each Kafka partition. They allow Kafka consumers to track processing progress, recover from failures, and resume from a known position after a restart. Since Kafka retains records independently of consumer offsets, we can also reset offsets to replay and reprocess historical data. Consumer-group offsets are stored in Kafka's internal __consumer_offsets topic. Offset management is also fundamental to delivery semantics such as at-least-once and exactly-once processing.

auto.offset.reset is a Kafka consumer configuration that determines where the consumer should start when it has no valid committed offset, or when its committed offset is no longer available. 32. What is a consumer group?

A Kafka consumer group is a logical group of consumers that work together to consume records from a topic. Kafka assigns partitions among the consumers in the group, with each partition assigned to only one consumer at a time within that group. This enables parallel processing and horizontal scalability. If a consumer joins or leaves the group, Kafka performs a rebalance and redistributes partitions. Each consumer group maintains its own committed offsets, so multiple consumer groups can independently consume and process the same topic for different applications.

33. Why do we use consumer groups?

We use Kafka consumer groups to achieve parallel processing, horizontal scalability, and fault tolerance. Within a consumer group, Kafka distributes partitions among consumers so that each partition is processed by only one consumer at a time. If a consumer fails or joins the group, Kafka rebalances the partition assignments. Consumer groups also provide independent consumption because different groups maintain separate offsets and can process the same topic for different business purposes. This makes consumer groups essential for scalable and resilient streaming architectures.

34. Can two consumers from the same consumer group consume the same partition simultaneously?

This ensures that consumers in the group share the workload rather than independently processing the same partition. However, consumers from different consumer groups can independently consume the same partition because each group maintains its own offsets.

35. Can two consumers from different consumer groups consume the same message?

36. What happens if a consumer crashes?

When a Kafka consumer crashes, Kafka detects that the consumer has left the consumer group and triggers a rebalance. The crashed consumer's partitions are reassigned to other consumers in the same group. The new consumer resumes processing based on the consumer group's committed offset for those partitions. If records were processed but their offsets were not committed before the crash, those records may be processed again, which is why Kafka consumers commonly use at-least-once processing and downstream idempotency where required.

37. What happens when a new consumer joins a consumer group?

When a new consumer joins a Kafka consumer group, the group membership changes and Kafka triggers a rebalance. The group coordinator and assignment mechanism redistribute the topic partitions among the consumers. The new consumer receives its assigned partitions and resumes from the appropriate committed offsets. This allows Kafka to scale consumption horizontally, but the maximum active consumer parallelism within a group is limited by the number of partitions.

38. What happens when the number of consumers is greater than the number of partitions?

If the number of consumers in a consumer group is greater than the number of partitions, some consumers remain idle because each partition can be assigned to only one consumer within a group. Therefore, the maximum active consumer parallelism is equal to the number of partitions. Extra consumers don't increase throughput; they mainly provide potential failover capacity

39. What happens when the number of consumers is less than the number of partitions?

When the number of consumers is less than the number of partitions, Kafka assigns multiple partitions to each consumer within the consumer group. This allows each consumer to process multiple partitions in parallel. If a consumer fails, Kafka triggers a rebalance and redistributes its partitions among the remaining consumers. The exact partition distribution depends on the assignment strategy.
