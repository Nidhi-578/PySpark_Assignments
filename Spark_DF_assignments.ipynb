3.Read orders.csv using an explicitly defined schema.

```python
from pyspark.sql.types import (StructType,StructField, StringType,LongType, IntegerType,DoubleType,DateType)

orders_schema=StructType([
    StructField("order_id",LongType(),True),
    StructField("customer_id",IntegerType(),True),
    StructField("order_date",DateType(),True),
    StructField("order_status",DoubleType(),True),
    StructField("payment_method",StringType(),True),
    StructField("shipping_city",DoubleType(),True),
    StructField("shipping_state",StringType(),True),
    StructField("total_amount",DoubleType(),True),
    StructField("discount_amount", DoubleType(), True),
    StructField("shipping_cost", DoubleType(), True)
])
orders=spark.read\
      .option("header",True)\
      .schema(orders_schema)\
      .csv(f"{base_path}/orders.csv")
orders.show(5)
orders.printSchema()

```

Assignment 7 — Window Functions
Using orders:

1. Find the customer's first order.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number
window_spec = Window.partitionBy("customer_id").orderBy("order_date")
first_orders = orders.withColumn(
    "order_rank",
    row_number().over(window_spec)
).filter(
    "order_rank = 1"
)
first_orders.show(5)
```

2.Find the customer's latest order.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, col

window_spec = Window.partitionBy("customer_id").orderBy(
    col("order_date").desc()
)
latest_orders = orders.withColumn(
    "order_rank",
    row_number().over(window_spec)
).filter(
    col("order_rank") == 1
)
latest_orders.show(5)
```

3.  Rank customers based on total spending.

```python
from pyspark.sql.functions import sum, rank
from pyspark.sql.window import Window

customer_spending = orders.groupBy("customer_id") \
    .agg(sum("total_amount").alias("total_spending"))
window_spec = Window.orderBy(customer_spending["total_spending"].desc())
ranked_customers = customer_spending.withColumn("customer_rank",rank().over(window_spec))
ranked_customers.show(5)
```

4. Rank orders within each customer.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank, col

window_spec = Window.partitionBy("customer_id").orderBy(col("total_amount").desc())
ranked_orders = orders.withColumn("order_rank",rank().over(window_spec))
ranked_orders.select("customer_id","order_id","total_amount","order_rank").show(5)
```

5. Find the top 3 orders for every customer.

```python
window_spec = Window.partitionBy("customer_id").orderBy(col("total_amount").desc())
top_3_orders = orders.withColumn(
    "order_rank",
    row_number().over(window_spec)
).filter(
    col("order_rank") <= 3
)

top_3_orders.select(
    "customer_id",
    "order_id",
    "total_amount",
    "order_rank"
).show(10)
```

6.Calculate running revenue for each customer.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import sum, col

window_spec = Window.partitionBy("customer_id").orderBy("order_date").rowsBetween(Window.unboundedPreceding, Window.currentRow)
running_revenue = orders.withColumn("running_revenue",sum("total_amount").over(window_spec))
running_revenue.select(
    "customer_id",
    "order_id",
    "order_date",
    "total_amount",
    "running_revenue"
).show(10)
```

7. Calculate previous order amount using lag().

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lag, col
window_spec = Window.partitionBy("customer_id").orderBy("order_date")
orders_with_previous = orders.withColumn("previous_order_amount",lag("total_amount").over(window_spec))

orders_with_previous.select(
    "customer_id",
    "order_id",
    "order_date",
    "total_amount",
    "previous_order_amount"
).show(10)
```

8. Calculate next order amount using lead().

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lead
window_spec = Window.partitionBy("customer_id").orderBy("order_date")
orders_with_next = orders.withColumn("next_order_amount",lead("total_amount").over(window_spec))

orders_with_next.select(
    "customer_id",
    "order_id",
    "order_date",
    "total_amount",
    "next_order_amount"
).show(10)
```

Calculate difference between current and previous order.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lag, col
window_spec = Window.partitionBy("customer_id").orderBy("order_date")
orders_with_difference = orders.withColumn("previous_order_amount",lag("total_amount").over(window_spec)
).withColumn(
    "order_amount_difference",
    col("total_amount") - col("previous_order_amount")
)

orders_with_difference.select(
    "customer_id",
    "order_id",
    "order_date",
    "total_amount",
    "previous_order_amount",
    "order_amount_difference"
).show(10)
```

10. Calculate cumulative order count for each customer.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import count
window_spec = Window.partitionBy("customer_id") \
    .orderBy("order_date") \
    .rowsBetween(Window.unboundedPreceding, Window.currentRow)

cumulative_orders = orders.withColumn(
    "cumulative_order_count",
    count("*").over(window_spec)
)

cumulative_orders.select(
    "customer_id",
    "order_id",
    "order_date",
    "cumulative_order_count"
).show(10)
```

Monthly Sales Report
Build a monthly business report.
Requirements
For every month calculate:
month
total_orders
completed_orders
cancelled_orders
total_revenue
total_discount
average_order_value
unique_customers
new_customers
Then calculate:
previous_month_revenue
revenue_difference
revenue_growth_percentage

```python
from pyspark.sql.functions import (
    col, count, sum, avg, countDistinct,
    when, date_format, lag, round
)
from pyspark.sql.window import Window

# 1. Create monthly report
monthly_report = orders.groupBy(
    date_format("order_date", "yyyy-MM").alias("month")
).agg(
    count("order_id").alias("total_orders"),

    sum(
        when(col("order_status") == "Completed", 1).otherwise(0)
    ).alias("completed_orders"),

    sum(
        when(col("order_status") == "Cancelled", 1).otherwise(0)
    ).alias("cancelled_orders"),

    sum("total_amount").alias("total_revenue"),

    sum("discount_amount").alias("total_discount"),

    avg("total_amount").alias("average_order_value"),

    countDistinct("customer_id").alias("unique_customers")
)

# 2. Calculate new customers

first_order = orders.groupBy("customer_id").agg(
    date_format(
        __import__("pyspark.sql.functions", fromlist=["min"]).min("order_date"),
        "yyyy-MM"
    ).alias("first_order_month")
)

new_customers = first_order.groupBy("first_order_month").agg(
    count("*").alias("new_customers")
)

# 3. Add new customers to monthly report
monthly_report = monthly_report.join(
    new_customers,
    monthly_report.month == new_customers.first_order_month,
    "left"
).drop("first_order_month")

# 4. Replace NULL new customer count with 0
monthly_report = monthly_report.fillna({
    "new_customers": 0
})

# 5. Compare with previous month's revenue
window_spec = Window.orderBy("month")

monthly_report = monthly_report.withColumn(
    "previous_month_revenue",
    lag("total_revenue").over(window_spec)
)

# 6. Revenue difference
monthly_report = monthly_report.withColumn(
    "revenue_difference",
    col("total_revenue") - col("previous_month_revenue")
)

# 7. Revenue growth percentage
monthly_report = monthly_report.withColumn(
    "revenue_growth_percentage",
    round(
        (
            (col("total_revenue") - col("previous_month_revenue"))
            / col("previous_month_revenue")
        ) * 100,
        2
    )
)

# 8. Final output
monthly_report.select(
    "month",
    "total_orders",
    "completed_orders",
    "cancelled_orders",
    "total_revenue",
    "total_discount",
    "average_order_value",
    "unique_customers",
    "new_customers",
    "previous_month_revenue",
    "revenue_difference",
    "revenue_growth_percentage"
).orderBy("month").show(10, truncate=False)
```

Additional requirements
1. Sort by month.
2. Find the highest-revenue month.
3. Find the lowest-revenue month.
4. Calculate cumulative revenue.
5. Write the report to Parquet.

```python
from pyspark.sql.functions import col, sum
from pyspark.sql.window import Window

# 1. Sort by month
monthly_report = monthly_report.orderBy("month")


# 2. Find the highest-revenue month
print("Highest Revenue Month:")

monthly_report.orderBy(
    col("total_revenue").desc()
).select(
    "month",
    "total_revenue"
).show(1)


# 3. Find the lowest-revenue month
print("Lowest Revenue Month:")

monthly_report.orderBy(
    col("total_revenue").asc()
).select(
    "month",
    "total_revenue"
).show(1)


# 4. Calculate cumulative revenue
cumulative_window = Window.orderBy("month") \
    .rowsBetween(
        Window.unboundedPreceding,
        Window.currentRow
    )

monthly_report = monthly_report.withColumn(
    "cumulative_revenue",
    sum("total_revenue").over(cumulative_window)
)


# Sort final report by month
monthly_report = monthly_report.orderBy("month")


# Display final report
monthly_report.show(10, truncate=False)


# 5. Write report to Parquet
output_path = f"{base_path}/monthly_sales_report"

monthly_report.write \
    .mode("overwrite") \
    .parquet(output_path)

print("Monthly Sales Report written to Parquet successfully.")
```

```python
#1. Load the DataFrames
from pyspark.sql.functions import (
    col, min, max, count, countDistinct,
    datediff, lag, months_between,
    date_format, add_months, current_date,
    when, round
)
from pyspark.sql.window import Window

customers = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/customers.csv")

orders = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/orders.csv")
```

```python
order_window = Window.partitionBy("customer_id").orderBy("order_date")
customer_orders = orders.withColumn(
    "order_number",
    row_number().over(order_window)
)
second_order = customer_orders.filter(
    col("order_number") == 2
).select(
    "customer_id",
    col("order_date").alias("second_order_date")
)
second_order.show(5)
```

Calculate days between first and second orders

```python
from pyspark.sql.functions import col, min, datediff, row_number
from pyspark.sql.window import Window

# 1. First order date for each customer
first_order = orders.groupBy("customer_id").agg(
    min("order_date").alias("first_order_date")
)

# 2. Number each customer's orders
order_window = Window.partitionBy("customer_id").orderBy("order_date")

customer_orders = orders.withColumn(
    "order_number",
    row_number().over(order_window)
)

# 3. Second order date
second_order = customer_orders.filter(
    col("order_number") == 2
).select(
    "customer_id",
    col("order_date").alias("second_order_date")
)

# 4. Join first and second orders
first_second = first_order.join(
    second_order,
    "customer_id",
    "inner"
)

# 5. Calculate days between first and second orders
first_second = first_second.withColumn(
    "days_between_orders",
    datediff(
        col("second_order_date"),
        col("first_order_date")
    )
)

# Display result
first_second.show(5, truncate=False)
```

Find customers who ordered in consecutive months.

```python
monthly_customer_orders = orders.select(
    "customer_id",
    date_format("order_date", "yyyy-MM").alias("order_month")
).distinct()

window_spec = Window.partitionBy(
    "customer_id"
).orderBy("order_month")

consecutive_month_customers = monthly_customer_orders.withColumn(
    "previous_month",
    lag("order_month").over(window_spec)
).filter(
    months_between(
        col("order_month"),
        col("previous_month")
    ) == 1
)

consecutive_month_customers.show(5)
```

Calculate monthly active customers.

```python
monthly_active_customers = orders.groupBy(
    date_format("order_date", "yyyy-MM").alias("month")
).agg(
    countDistinct("customer_id").alias("monthly_active_customers")
).orderBy("month")

monthly_active_customers.show(5)
```

Calculate customers retained from previous month.

```python
customer_months = orders.select(
    "customer_id",
    date_format("order_date", "yyyy-MM").alias("month")
).distinct()

current_month = customer_months.alias("current")

previous_month = customer_months.alias("previous")

retained_customers = current_month.join(
    previous_month,
    (col("current.customer_id") == col("previous.customer_id")) &
    (
        months_between(
            col("current.month"),
            col("previous.month")
        ) == 1
    ),
    "inner"
).select(
    col("current.month").alias("month"),
    col("current.customer_id")
).distinct()

retained_summary = retained_customers.groupBy(
    "month"
).agg(
    countDistinct("customer_id").alias("retained_customers")
).orderBy("month")

retained_summary.show(5)
```

```python
active_with_previous = monthly_active_customers.withColumn(
    "previous_month_active_customers",
    lag("monthly_active_customers").over(
        Window.orderBy("month")
    )
)

retention_report = active_with_previous.join(
    retained_summary,
    "month",
    "left"
).fillna({
    "retained_customers": 0
}).withColumn(
    "retention_percentage",
    round(
        col("retained_customers") * 100.0 /
        col("previous_month_active_customers"),
        2
    )
).orderBy("month")

retention_report.show(5, truncate=False)
```

Data Quality Assignment

1. Identify NULL records.
2. Identify duplicates.
3. Identify invalid emails.
4. Identify negative amounts.
5. Identify invalid dates.
6. Standardize text fields.
7. Remove invalid records.
8. Create a data-quality status column.
9. Generate a rejected-record dataset.
10. Generate a cleaned dataset.
Final outputs:
clean_orders/
rejected_orders/
data_quality_report/

```python
from pyspark.sql.functions import (
    col, trim, lower, regexp_extract,
    to_date, when, lit, count
)
from pyspark.sql.window import Window

# 1. Standardize text fields

cleaning_df = orders \
    .withColumn("order_status", trim(lower(col("order_status")))) \
    .withColumn("payment_method", trim(lower(col("payment_method")))) \
    .withColumn("shipping_city", trim(col("shipping_city"))) \
    .withColumn("shipping_state", trim(col("shipping_state")))

# 2. Identify NULL records

null_condition = (
    col("order_id").isNull() |
    col("customer_id").isNull() |
    col("order_date").isNull() |
    col("order_status").isNull() |
    col("total_amount").isNull()
)

# 3. Identify negative amounts

negative_amount_condition = (
    (col("total_amount") < 0) |
    (col("discount_amount") < 0) |
    (col("shipping_cost") < 0)
)

# 4. Identify invalid dates

invalid_date_condition = col("order_date").isNull()

# 5. Identify duplicate orders


duplicate_window = Window.partitionBy("order_id")

cleaning_df = cleaning_df.withColumn(
    "duplicate_count",
    count("*").over(duplicate_window)
)

duplicate_condition = col("duplicate_count") > 1


# 6. Create data-quality status


cleaning_df = cleaning_df.withColumn(
    "data_quality_status",
    when(null_condition, "INVALID_NULL")
    .when(duplicate_condition, "INVALID_DUPLICATE")
    .when(negative_amount_condition, "INVALID_NEGATIVE_AMOUNT")
    .when(invalid_date_condition, "INVALID_DATE")
    .otherwise("VALID")
)

# 7. Generate rejected records

rejected_orders = cleaning_df.filter(
    col("data_quality_status") != "VALID"
).drop("duplicate_count")

# 8. Generate cleaned records

clean_orders = cleaning_df.filter(
    col("data_quality_status") == "VALID"
).drop("duplicate_count")

# 9. Display results

print("Rejected Orders:")
rejected_orders.show(5, truncate=False)

print("Clean Orders:")
clean_orders.show(5, truncate=False)

# 10. Data Quality Report


data_quality_report = cleaning_df.groupBy(
    "data_quality_status"
).agg(
    count("*").alias("record_count")
).orderBy("data_quality_status")

print("Data Quality Report:")
data_quality_report.show()

# 11. Write final outputs as Parquet


clean_orders.write \
    .mode("overwrite") \
    .parquet(f"{base_path}/clean_orders")

rejected_orders.write \
    .mode("overwrite") \
    .parquet(f"{base_path}/rejected_orders")

data_quality_report.write \
    .mode("overwrite") \
    .parquet(f"{base_path}/data_quality_report")

print("All outputs written successfully.")
```

Tasks
Create a pipeline that:
1. Reads all datasets.
2. Filters unnecessary records.
3. Joins orders with customers.
4. Joins order items with products.
5. Creates revenue metrics.
6. Calculates customer-level revenue.
7. Calculates product-level revenue.
8. Calculates category-level revenue.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum
from pyspark import StorageLevel

# 1. Read all datasets

orders = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/orders.csv")

order_items = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/order_items.csv")

customers = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/customers.csv")

products = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv(f"{base_path}/products.csv")

# 2. Filter unnecessary records and columns


orders_filtered = orders.filter(
    col("order_status") == "Completed"
).select(
    "order_id",
    "customer_id",
    "order_date",
    "total_amount",
    "discount_amount"
)

order_items_filtered = order_items.select(
    "order_id",
    "product_id",
    "quantity"
)

customers_filtered = customers.select(
    "customer_id",
    "first_name",
    "last_name",
    "city",
    "state"
)

products_filtered = products.select(
    "product_id",
    "product_name",
    "category",
    "selling_price",
    "unit_cost"
)


# 3. Repartition large datasets by join keys


orders_filtered = orders_filtered.repartition("customer_id")

order_items_filtered = order_items_filtered.repartition("product_id")



# 4. Cache orders because it is reused


orders_filtered.cache()

# Materialize cache
orders_filtered.count()

# 5. Join orders with customers


customer_orders = orders_filtered.join(
    customers_filtered,
    "customer_id",
    "inner"
)

# 6. Join order_items with products

product_items = order_items_filtered.join(
    products_filtered,
    "product_id",
    "inner"
)

# 7. Create revenue metrics


product_items = product_items.withColumn(
    "revenue",
    col("quantity") * col("selling_price")
).withColumn(
    "profit",
    col("quantity") *
    (col("selling_price") - col("unit_cost"))
)

# 8. Persist product-level data


product_items.persist(StorageLevel.MEMORY_AND_DISK)

# Materialize persist
product_items.count()


# --------------------------------------------------
# 9. Customer-level revenue
# --------------------------------------------------

customer_revenue = customer_orders.join(
    product_items.select(
        "order_id",
        "revenue"
    ),
    "order_id",
    "inner"
).groupBy(
    "customer_id",
    "first_name",
    "last_name"
).agg(
    sum("revenue").alias("customer_revenue")
)

# 10. Product-level revenue


product_revenue = product_items.groupBy(
    "product_id",
    "product_name",
    "category"
).agg(
    sum("revenue").alias("product_revenue")
)

# 11. Category-level revenue


category_revenue = product_items.groupBy(
    "category"
).agg(
    sum("revenue").alias("category_revenue")
)


# 12. Reduce partitions before writing small outputs


customer_revenue = customer_revenue.coalesce(4)

product_revenue = product_revenue.coalesce(4)

category_revenue = category_revenue.coalesce(2)


# 13. Display results


print("Customer Revenue:")
customer_revenue.show(10, truncate=False)

print("Product Revenue:")
product_revenue.show(10, truncate=False)

print("Category Revenue:")
category_revenue.show(10, truncate=False)


# 14. Remove cached data


orders_filtered.unpersist()
product_items.unpersist()
```

Step 3 — Transformation
Standardize:

Create required derived columns.

```python
from pyspark.sql.functions import (
    col, trim, initcap, lower, upper, concat_ws, datediff, current_date, floor,
    months_between, year, month, quarter, when )


# Standardize CUSTOMER data

customers_transformed = customers \
    .withColumn(
        "first_name",
        initcap(trim(col("first_name")))
    ) \
    .withColumn(
        "last_name",
        initcap(trim(col("last_name")))
    ) \
    .withColumn(
        "email",
        lower(trim(col("email")))
    ) \
    .withColumn(
        "city",
        initcap(trim(col("city")))
    ) \
    .withColumn(
        "state",
        initcap(trim(col("state")))
    )

# Create full customer name

customers_transformed = customers_transformed.withColumn(
    "customer_name",
    concat_ws(
        " ",
        col("first_name"),
        col("last_name")
    )
)

# Create customer age

customers_transformed = customers_transformed.withColumn(
    "customer_age",
    floor(
        months_between(
            current_date(),
            col("date_of_birth")
        ) / 12
    )
)

# Create registration year and month

customers_transformed = customers_transformed \
    .withColumn(
        "registration_year",
        year("registration_date")
    ) \
    .withColumn(
        "registration_month",
        month("registration_date")
    )

# Standardize ORDERS


orders_transformed = orders \
    .withColumn(
        "order_status",
        initcap(trim(col("order_status")))
    ) \
    .withColumn(
        "payment_method",
        upper(trim(col("payment_method")))
    ) \
    .withColumn(
        "shipping_city",
        initcap(trim(col("shipping_city")))
    ) \
    .withColumn(
        "shipping_state",
        initcap(trim(col("shipping_state")))
    )

#  Create order-level derived columns


orders_transformed = orders_transformed \
    .withColumn(
        "net_amount",
        col("total_amount") - col("discount_amount")
    ) \
    .withColumn(
        "order_year",
        year("order_date")
    ) \
    .withColumn(
        "order_month",
        month("order_date")
    ) \
    .withColumn(
        "order_quarter",
        quarter("order_date")
    )

# Display results

print("TRANSFORMED CUSTOMERS")
customers_transformed.select(
    "customer_id",
    "customer_name",
    "email",
    "city",
    "state",
    "customer_age",
    "registration_year",
    "registration_month"
).show(5, truncate=False)


print("TRANSFORMED ORDERS")
orders_transformed.select(
    "order_id",
    "customer_id",
    "order_date",
    "order_status",
    "payment_method",
    "total_amount",
    "discount_amount",
    "net_amount",
    "order_year",
    "order_month",
    "order_quarter"
).show(5, truncate=False)
```

Step 4 — Join
Build an order-level dataset.

```python
from pyspark.sql.functions import (
    col,
    sum,
    countDistinct,
    concat_ws
)
# Prepare product information

order_item_product = order_items.join(
    products.select(
        "product_id",
        "product_name",
        "category",
        "brand",
        "selling_price",
        "unit_cost"
    ),
    "product_id",
    "inner"
)
# Calculate order-level product metrics

order_product_metrics = order_item_product.groupBy(
    "order_id"
).agg(
    sum("quantity").alias("total_quantity"),

    sum(
        col("quantity") * col("selling_price")
    ).alias("product_revenue"),

    sum(
        col("quantity") *
        (col("selling_price") - col("unit_cost"))
    ).alias("product_profit"),

    countDistinct("product_id").alias("unique_products")
)

# Join ORDERS with CUSTOMERS

order_customer = orders_transformed.join(
    customers_transformed.select(
        "customer_id",
        "customer_name",
        "email",
        "city",
        "state",
        "registration_date",
        "customer_segment"
    ),
    "customer_id",
    "left"
)


# Join order/customer data with product metrics

order_level_dataset = order_customer.join(
    order_product_metrics,
    "order_id",
    "left"
)

# Select final order-level columns

order_level_dataset = order_level_dataset.select(
    "order_id",
    "customer_id",
    "customer_name",
    "email",
    "city",
    "state",
    "registration_date",
    "customer_segment",
    "order_date",
    "order_status",
    "payment_method",
    "total_amount",
    "discount_amount",
    "shipping_cost",
    "net_amount",
    "total_quantity",
    "product_revenue",
    "product_profit",
    "unique_products",
    "order_year",
    "order_month",
    "order_quarter"
)

# Display final order-level dataset

order_level_dataset.show(
    5,
    truncate=False
)

print("Order-level columns:")
order_level_dataset.printSchema()
```

Step 5 — Aggregation
Create:
daily_sales

```python
from pyspark.sql.functions import (
    col,
    sum,
    count,
    countDistinct,
    avg,
    year,
    month,
    date_format
)

# DAILY SALES

daily_sales = order_level_dataset.groupBy(
    "order_date"
).agg(
    countDistinct("order_id").alias("total_orders"),
    sum("total_amount").alias("total_revenue"),
    sum("discount_amount").alias("total_discount"),
    sum("shipping_cost").alias("total_shipping_cost"),
    sum("net_amount").alias("net_revenue")
).orderBy("order_date")

# MONTHLY SALES

monthly_sales = order_level_dataset.groupBy(
    date_format("order_date", "yyyy-MM").alias("month")
).agg(
    countDistinct("order_id").alias("total_orders"),
    sum("total_amount").alias("total_revenue"),
    sum("discount_amount").alias("total_discount"),
    sum("shipping_cost").alias("total_shipping_cost"),
    sum("net_amount").alias("net_revenue")
).orderBy("month")

# CUSTOMER SALES

customer_sales = order_level_dataset.groupBy(
    "customer_id",
    "customer_name"
).agg(
    countDistinct("order_id").alias("total_orders"),
    sum("total_amount").alias("total_revenue"),
    sum("discount_amount").alias("total_discount"),
    sum("net_amount").alias("net_revenue"),
    avg("total_amount").alias("average_order_value")
).orderBy(
    col("total_revenue").desc()
)

# PRODUCT SALES

# Aggregate order_items with products first
product_sales = order_items.join(
    products.select(
        "product_id",
        "product_name",
        "category",
        "brand",
        "selling_price",
        "unit_cost"
    ),
    "product_id",
    "inner"
).groupBy(
    "product_id",
    "product_name",
    "category",
    "brand"
).agg(
    sum("quantity").alias("quantity_sold"),

    sum(
        col("quantity") * col("selling_price")
    ).alias("total_revenue"),

    sum(
        col("quantity") *
        (col("selling_price") - col("unit_cost"))
    ).alias("total_profit")
).orderBy(
    col("total_revenue").desc()
)

# CATEGORY SALES

category_sales = product_sales.groupBy(
    "category"
).agg(
    sum("quantity_sold").alias("quantity_sold"),
    sum("total_revenue").alias("total_revenue"),
    sum("total_profit").alias("total_profit")
).orderBy(
    col("total_revenue").desc()
)

# DISPLAY RESULTS

print("DAILY SALES")
daily_sales.show(5, truncate=False)

print("MONTHLY SALES")
monthly_sales.show(5, truncate=False)

print("CUSTOMER SALES")
customer_sales.show(5, truncate=False)

print("PRODUCT SALES")
product_sales.show(5, truncate=False)

print("CATEGORY SALES")
category_sales.show(5, truncate=False)
```

Step 6 — Window Analysis

```python
from pyspark.sql.functions import (col,sum,row_number,rank,lag,min,max)
from pyspark.sql.window import Window

# CUSTOMER RANK
# Rank customers based on total revenue

customer_rank_window = Window.orderBy(
    col("total_revenue").desc()
)

customer_sales = customer_sales.withColumn(
    "customer_rank",
    rank().over(customer_rank_window)
)

# PRODUCT RANK
# Rank products based on total revenue

product_rank_window = Window.orderBy(
    col("total_revenue").desc()
)

product_sales = product_sales.withColumn(
    "product_rank",
    rank().over(product_rank_window)
)

# PREVIOUS MONTH SALES
previous_month_window = Window.orderBy("month")

monthly_sales = monthly_sales.withColumn(
    "previous_month_sales",
    lag("total_revenue").over(previous_month_window)
)

# RUNNING SALES

running_sales_window = Window.orderBy("month") \
    .rowsBetween(
        Window.unboundedPreceding,
        Window.currentRow
    )

monthly_sales = monthly_sales.withColumn(
    "running_sales",
    sum("total_revenue").over(running_sales_window)
)

#  FIRST ORDER
# Find first order date for each customer


customer_order_window = Window.partitionBy(
    "customer_id"
)

customer_orders_analysis = order_level_dataset.withColumn(
    "first_order",
    min("order_date").over(customer_order_window)
)

# LATEST ORDER
# Find latest order date for each customer

customer_orders_analysis = customer_orders_analysis.withColumn(
    "latest_order",
    max("order_date").over(customer_order_window)
)

# DISPLAY RESULTS
print("CUSTOMER RANK")
customer_sales.select(
    "customer_id",
    "customer_name",
    "total_revenue",
    "customer_rank"
).show(5, truncate=False)


print("PRODUCT RANK")
product_sales.select(
    "product_id",
    "product_name",
    "category",
    "total_revenue",
    "product_rank"
).show(5, truncate=False)


print("MONTHLY SALES WITH WINDOW ANALYSIS")
monthly_sales.show(5, truncate=False)


print("CUSTOMER FIRST AND LATEST ORDER")
customer_orders_analysis.select(
    "customer_id",
    "order_id",
    "order_date",
    "first_order",
    "latest_order"
).show(5, truncate=False)
```

Step 7 — UDF
Create at least one meaningful UDF.
Example:
customer risk classification

```python
from pyspark.sql.functions import (
    col,
    datediff,
    current_date,
    udf
)
from pyspark.sql.types import StringType

# Create Python function

def customer_risk(days_since_last_order):

    if days_since_last_order is None:
        return "High Risk"

    elif days_since_last_order > 180:
        return "High Risk"

    elif days_since_last_order > 90:
        return "Medium Risk"

    else:
        return "Low Risk"

# Convert Python function to Spark UDF

customer_risk_udf = udf(
    customer_risk,
    StringType()
)

# Calculate days since latest order

customer_risk_df = customer_orders_analysis.withColumn(
    "days_since_last_order",
    datediff(
        current_date(),
        col("latest_order")
    )
)

# Apply UDF
customer_risk_df = customer_risk_df.withColumn(
    "customer_risk",
    customer_risk_udf(
        col("days_since_last_order")
    )
)

# Display result

customer_risk_df.select(
    "customer_id",
    "customer_name",
    "first_order",
    "latest_order",
    "days_since_last_order",
    "customer_risk"
).dropDuplicates(
    ["customer_id"]
).show(5, truncate=False)
```

```python
from pyspark import StorageLevel
from pyspark.sql.functions import col

# CACHE

monthly_sales.cache()

# Materialize the cache
monthly_sales.count()

print("monthly_sales cached successfully.")

# PERSIST

order_level_dataset.persist(
    StorageLevel.MEMORY_AND_DISK
)

# Materialize persistence
order_level_dataset.count()

print("order_level_dataset persisted successfully.")

# REPARTITION
# Repartition customer-level data by customer_id.
# This can help when subsequent operations use customer_id.

customer_sales_optimized = customer_sales.repartition(
    "customer_id"
)

print(
    "Customer sales partitions:",
    customer_sales_optimized.rdd.getNumPartitions()
)
# REPARTITION PRODUCT DATA

# Product analysis commonly uses product_id.

product_sales_optimized = product_sales.repartition(
    "product_id"
)

print(
    "Product sales partitions:",
    product_sales_optimized.rdd.getNumPartitions()
)

# 5. COALESCE
# Reduce the number of partitions before writing smaller
# aggregated datasets.

category_sales_optimized = category_sales.coalesce(2)

print(
    "Category sales partitions:",
    category_sales_optimized.rdd.getNumPartitions()
)

# Display optimized results

print("\nCustomer Sales:")
customer_sales_optimized.show(5, truncate=False)

print("\nProduct Sales:")
product_sales_optimized.show(5, truncate=False)

print("\nCategory Sales:")
category_sales_optimized.show(5, truncate=False)

# Remove cached/persisted DataFrames when finished

monthly_sales.unpersist()

order_level_dataset.unpersist()

print("\nOptimization step completed.")
```

1.** Why did you choose these partition columns?**
daily_sales → year and month: Sales data is time-based, so queries commonly filter by year or month.
product_sales → category: Category is a useful business dimension for product analysis.
customer_sales was not partitioned because it is already an aggregated customer-level dataset and did not have a suitable partition column.

**2. Where did shuffle happen?**

Shuffle can happen when Spark needs to move data between partitions.

In our pipeline, shuffle can occur during:

groupBy()
join()
orderBy()
distinct()
repartition()
Some window functions

orders.groupBy("customer_id").agg(...)
Spark may need to bring records with the same customer_id into the same partition.

**3. Which joins caused shuffle?**

The important joins were:

orders.join(customers, "customer_id")

and:

order_items.join(products, "product_id")

These joins may require shuffle when the data is not already appropriately partitioned on the join keys.

The exact physical execution can be checked using:

df.explain("formatted")

4**. Why did you use cache/persist?**

I used cache() when a DataFrame was reused multiple times.

I used:

order_level_dataset.persist(StorageLevel.MEMORY_AND_DISK)

because the order-level dataset is a relatively large intermediate DataFrame that can be reused by multiple downstream operations.

Caching/persisting avoids repeatedly recalculating the same expensive transformations.

**5. Where could cache hurt performance?**

Caching can hurt when:

The DataFrame is used only once.
The DataFrame is very large.
Available memory is insufficient.
Too many DataFrames are cached.
Cached data causes memory pressure and eviction.
The cost of creating the cache is greater than the computation saved.

Therefore, caching should be used selectively.

6.** Why did you use a UDF?**

I created a customer risk classification UDF based on the number of days since the customer's latest order.

The classification was:

0–90 days       → Low Risk
91–180 days     → Medium Risk
More than 180   → High Risk
NULL            → High Risk

This demonstrates how a custom business rule can be implemented as a reusable function.

7. Could the UDF be replaced with native Spark functions?

Yes.

The same logic can be implemented using when() and otherwise():

customer_risk_df = customer_risk_df.withColumn(
    "customer_risk",
    when(
        col("days_since_last_order").isNull(),
        "High Risk"
    ).when(
        col("days_since_last_order") > 180,
        "High Risk"
    ).when(
        col("days_since_last_order") > 90,
        "Medium Risk"
    ).otherwise(
        "Low Risk"
    )
)

For this type of logic, native Spark functions are generally preferable because Spark can optimize them better than a Python UDF.


9. Why did you use repartition() or coalesce()?

I used repartition() when I wanted to redistribute data based on a useful key:

customer_sales.repartition("customer_id")

repartition() performs a shuffle but can improve data distribution for subsequent operations.

I used coalesce() when I wanted to reduce the number of partitions, particularly for smaller aggregated outputs:

category_sales.coalesce(2)

coalesce() generally avoids a full shuffle and is therefore useful for reducing partitions.

10. What transformations are narrow and wide?
Narrow transformations

Narrow transformations generally do not require data to move between partitions.

Examples:

filter()
where()
select()
withColumn()
withColumnRenamed()

Example:

orders.filter(col("order_status") == "Completed")
Wide transformations

Wide transformations can require data to move between partitions and therefore can cause shuffle.

Examples:

groupBy()
join()
orderBy()
distinct()
repartition()

Window operations can also require redistribution depending on the partitioning and ordering.

For example:

orders.groupBy("customer_id").agg(
    sum("total_amount")
)

Spark needs to bring records belonging to the same grouping key together, which can require a shuffle.
