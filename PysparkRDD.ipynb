Q1
Dataset
[5, 10, 15, 20, 25, 30]
Problem:
Create an RDD from the dataset and multiply every number by 2.
Expected Output
[10, 20, 30, 40, 50, 60]

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Solution").getOrCreate()
sc=spark.sparkContext
numbers=[5, 10, 15, 20, 25, 30]
rdd=sc.parallelize(numbers)
result=rdd.map(lambda x:x*2)
print(result.collect())
```

Q50
Problem:
Perform the following complete analysis:
1.Calculate revenue for every order.
2.Keep only orders with revenue greater than 50000.
3.Calculate total revenue for each product.
4.Find the product with the highest revenue.
5.Save the final result into a single output partition.
Expected Output
Orders retained:
O1 -> Laptop -> 100000
O2 -> Mobile -> 100000
O6 -> Laptop -> 200000
Product Revenue:
Laptop -> 300000
Mobile -> 100000
Highest Revenue Product:
Laptop -> 300000
Output:
1 part file

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName('RDDPractice').getOrCreate()
sc=spark.sparkContext
orders=[("O1", "Laptop", 2, 50000),
    ("O2", "Mobile", 5, 20000),
    ("O3", "Laptop", 1, 50000),
    ("O4", "Tablet", 3, 15000),
    ("O5", "Mobile", 2, 20000),
    ("O6", "Laptop", 4, 50000)]
rdd=sc.parallelize(orders)
# 1. Calculate revenue for every order
order_revenue = rdd.map(
    lambda x: (x[0], x[1], x[2] * x[3])
)

# 2. Keep only orders with revenue > 50000
filtered = order_revenue.filter(
    lambda x: x[2] > 50000
)

print("Orders retained:")
print(filtered.collect())

# 3. Calculate total revenue for each product
product_revenue = filtered \
    .map(lambda x: (x[1], x[2])) \
    .reduceByKey(lambda x, y: x + y)

print("Product Revenue:")
print(product_revenue.collect())

# 4. Find product with highest revenue
highest = product_revenue.reduce(
    lambda x, y: x if x[1] > y[1] else y
)

print("Highest Revenue Product:")
print(highest)

# 5. Save final result into a single output partition
final_result = sc.parallelize([highest]).coalesce(1)
final_result.saveAsTextFile("output/highest_revenue")
```
